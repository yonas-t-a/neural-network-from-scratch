```python
import numpy as np
from nnfs.datasets import spiral_data
import nnfs

nnfs.init()
```

```python
class Layer_Dense:
    def __init__(self, input, neuron):
        self.input = input 
        self.neuron = neuron
        self.weights = 0.01 * np.random.randn(input, neuron)  
        self.biases = np.zeros((1, neuron))
    
    def forward(self, input):
        
        # since the weights matrix is input X neuron we don't transpue the weights
        self.output =  np.dot(input, self.weights) + self.biases
```

```python
class Activation_ReLu:
    def forward(self, input):
        self.output = np.maximum(0, input)
```

```python
X,y = spiral_data(samples=100, classes=3)
```

```python
dense1 =  Layer_Dense(2,3)
activation = Activation_ReLu()
```

```python
dense1.forward(X)
activation.forward(dense1.output)
```
