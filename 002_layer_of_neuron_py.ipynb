```python
# 4 input, 3 neuron 
inputs = [1,2,3,2.5]
weights1 = [0.2, 0.8, -0.5, 1.0]
weights2 = [0.5, -0.91, 0.26, -0.5]
weights3 = [-0.26, 0.27, 0.17, 0.87]

bias1 = 2
bias2 = 3
bias3 = 0.5

output = [0,0,0]

for i, w in zip(inputs,weights1):
    output[0] += i*w
output[0]+= bias1

for i, w in zip(inputs, weights2):
    output[1] += i*w
output[1] += bias2
    
for i, w in zip(inputs, weights3):
    output[2] += i*w
output[2]+=bias3

print(output)
```

```python
#fully connected neural network 
inputs = [1,2,3,2.5]
weights = [  [0.2, 0.8, -0.5, 1.0],
            [0.5, -0.91, 0.26, -0.5],
            [-0.26, 0.27, 0.17, 0.87]   ]

biases = [2,3,0.5]

outputs = []

for w, b in zip(weights, biases):
    output = 0
    
    
    for i, n_w in zip(inputs, w):
        output += i *n_w
    
    output += b
    
    outputs.append(output)

print(outputs)
    
```
