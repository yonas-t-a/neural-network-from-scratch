```python
# 4 input, 3 neuron 
inputs = [1,2,3,2.5]
weight1 = [0,2, 0.8, -0.5, 1.0]
weight2 = [0.5, -0.91, 0.26, -0.5]
weight3 = [-0.26, 0.27, 0.17, 0.87]

b1 = 2
b2 = 3
b3 = 0.5

output = [0,0,0]

for i, w in zip(inputs,weight1):
    output[0] += i*w
output[0]+= b1

for i, w in zip(inputs, weight2):
    output[1] += i*w
output[1] += b2
    
for i, w in zip(inputs, weight3):
    output[2] += i*w
output[2]+=b3

print(output)
```

```python
#fully connected neural network 
inputs = [1,2,3,2.5]
weight = [  [0,2, 0.8, -0.5, 1.0],
            [0.5, -0.91, 0.26, -0.5],
            [-0.26, 0.27, 0.17, 0.87]   ]

biases = [2,3,0.5]

outputs = []

for w, b in zip(weight, biases):
    output = 0
    
    
    for i, n_w in zip(inputs, w):
        output += i *n_w
    
    output += b
    
    outputs.append(output)

print(outputs)
    
```
