```python
# neuron with 3 input
input = [1,2,3] 
weight = [0,2, 0.8, -0.5]
baise = 2


output = weight[0]*input[0] + weight[1]*input[1] + weight[2]*input[2] + baise

print(output)
```

```python
# neuron with 4 input 

input = [1,2,3, 2.5] 
weight = [0,2, 0.8, -0.5, 1.0]
baise = 2

output = 0

for i, w in zip(input, weight):
    output += i*w

output += baise

print(output)
```
