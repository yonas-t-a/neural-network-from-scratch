```python
# neuron with 3 inputs
inputs = [1,2,3] 
weights = [0.2, 0.8, -0.5]
bias = 2


output = weights[0]*inputs[0] + weights[1]*inputs[1] + weights[2]*inputs[2] + bias

print(output)
```

```python
# neuron with 4 inputs 

inputs = [1,2,3, 2.5] 
weights = [0.2, 0.8, -0.5, 1.0]
bias = 2

output = 0

for i, w in zip(inputs, weights):
    output += i*w

output += bias

print(output)
```
